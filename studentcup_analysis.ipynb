```python
import pandas as pd

df = pd.read_csv("data/train.csv")

df.isnull().sum().sort_values(ascending=False)
```

```python
cols = [
    "事業所数",
    "工場数",
    "店舗数",
    "資本金",
    "総資産",
    "流動資産",
    "固定資産",
    "負債",
    "短期借入金",
    "長期借入金",
    "純資産",
    "自己資本",
    "売上",
    "営業利益",
    "経常利益",
    "当期純利益",
    "営業CF",
    "減価償却費",
    "運転資本変動",
    "投資CF",
    "有形固定資産変動",
    "無形固定資産変動(ソフトウェア関連)",
    "アンケート１",
    "アンケート２",
    "アンケート３",
    "アンケート４",
    "アンケート５",
    "アンケート６",
    "アンケート７",
    "アンケート８",
    "アンケート９",
    "アンケート１０",
    "アンケート１１"
]

df[cols].describe()
```

```python
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(col, "外れ値数:", len(outliers))
```

```python
corr = df[cols + ["購入フラグ"]].corr()["購入フラグ"]

corr.abs().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
import japanize_matplotlib

fig, axes = plt.subplots(5, 7, figsize=(20, 14))

axes = axes.flatten()

for i, col in enumerate(cols):
    axes[i].hist(df[col].dropna(), bins=30)
    axes[i].set_title(col, fontsize=10)
    axes[i].tick_params(axis="x", labelrotation=45)

# 余った枠を消す
for i in range(len(cols), len(axes)):
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


# 5行×7列
fig, axes = plt.subplots(5, 7, figsize=(22, 15))

axes = axes.flatten()

for i, col in enumerate(cols):
    sns.histplot(
        df[col].dropna(),
        bins=30,
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(col, fontsize=10)
    axes[i].set_xlabel("")
    axes[i].set_ylabel("")

# 余った枠を非表示
for i in range(len(cols), len(axes)):
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(5, 7, figsize=(22, 15))

axes = axes.flatten()

for i, col in enumerate(cols):
    sns.boxplot(
        data=df,
        x="購入フラグ",
        y=col,
        ax=axes[i]
    )
    
    axes[i].set_title(col, fontsize=10)
    axes[i].set_xlabel("")
    axes[i].set_ylabel("")

# 余った枠を消す
for i in range(len(cols), len(axes)):
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(5, 7, figsize=(22, 15))

axes = axes.flatten()

for i, col in enumerate(cols):
    sns.violinplot(
        data=df,
        x="購入フラグ",
        y=col,
        ax=axes[i]
    )

    axes[i].set_title(col, fontsize=10)
    axes[i].set_xlabel("")
    axes[i].set_ylabel("")

# 余った場所を消す
for i in range(len(cols), len(axes)):
    axes[i].axis("off")

plt.tight_layout()
plt.show()
```
